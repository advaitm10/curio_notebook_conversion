```python
import pandas as pd
import numpy as np

df = pd.read_csv("/home/advaitm10/CPS.csv")
df
```

```python
HS_only = df[df["Elementary, Middle, or High School"] == 'HS']
```

```python
MS_only = df[df["Elementary, Middle, or High School"] == 'MS']
```

```python
ES_only = df[df["Elementary, Middle, or High School"] == 'ES']
```

```python
community_zipcode = HS_only[["Community Area Name", "ZIP Code"]]
grouped_data = community_zipcode.groupby("Community Area Name").size().reset_index(name="High School Count")
grouped_data
```

```python
community_zipcode = MS_only[["Community Area Name", "ZIP Code"]]
grouped_data = community_zipcode.groupby("Community Area Name").size().reset_index(name="Middle School Count")
grouped_data
```

```python
community_zipcode = ES_only[["Community Area Name", "ZIP Code"]]
grouped_data = community_zipcode.groupby("Community Area Name").size().reset_index(name="Elementary School Count")
grouped_data
```
